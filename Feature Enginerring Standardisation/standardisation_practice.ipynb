```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report , confusion_matrix , accuracy_score 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Social_Network_Ads.csv")
df = df.iloc[ : , 2 :]
print(df.head())
print(df.tail())
```

```python
#x = df.iloc[ : , 2 : 4]
#y = df.iloc[ : , 4]

x = df.drop('Purchased' , axis = 1)
y = df['Purchased']
x_train , x_test , y_train , y_test = train_test_split(x , y)
scalling = StandardScaler()
scalling.fit(x_train)
x_train_scalled = scalling.transform(x_train)
x_test_scalled = scalling.transform(x_test)

#print(x_train_scalled)
#print(x_test_scalled)


```

```python
# scalling_comparasion = pd.DataFrame({'Actual x_train_age' : x_train.iloc[ : , 1 ].values, 'Scalled x_train_age' : x_train_scalled[ : , 1]})
# print(scalling_comparasion.head())
x_train_scalled = pd.DataFrame(
    x_train_scalled,
    columns=x_train.columns       # Converting NUmpy Array into Dataframe to use iloc operation
)
x_test_scalled = pd.DataFrame(
    x_test_scalled , 
    columns = x_test.columns
)
# print(x_train_scalled.head())
# print(x_test_scalled.head())
scalling_comparasion = pd.DataFrame({'Actual x_train_age' : x_train.iloc[ : , 0 ],
                                     'Actual x_train_estimated_salary' : x_train.iloc[ : , 1], 
                                     'Scalled x_train_age' : x_train_scalled.iloc[ : , 0],
                                     'Scaled x_train_estimated_salary' : x_train_scalled.iloc[ : , 1]
                                     })

print(scalling_comparasion.head())
```

```python
model_1_non_scaled = LogisticRegression()
model_1_non_scaled.fit(x_train , y_train)
y_pred = model_1_non_scaled.predict(x_test)
y_pred = pd.DataFrame(
    y_pred ,
    columns = ['Purchased'],
    index = y_test.index
)
# print(y_train)
# print(y_pred)
checking_non_scaled_output = pd.DataFrame({'Actual output' : y_test.squeeze() , 'Non scalled Predicted output' : y_pred.squeeze()})
pd.set_option('display.max_rows' , None)
pd.set_option('display.max_columns' , None)
print(checking_non_scaled_output)
```

```python
print('classification report non scalled :-' , classification_report (y_test , y_pred))
print('confusion matrix non scalled :-' , confusion_matrix(y_test , y_pred))
print('accuracy_score non scalled :-' , accuracy_score(y_test , y_pred))
```

```python
model_2_scalled = LogisticRegression()
model_2_scalled.fit(x_train_scalled , y_train)

y_pred2 = model_2_scalled.predict(x_test_scalled)

y_pred2 = pd.DataFrame(
    y_pred2 ,
    columns = ['Purchased'],
    index = y_test.index
)

checking_scaled = pd.DataFrame({'Actual output' : y_test.squeeze() , 'scaled Predicted output' : y_pred2.squeeze()})
pd.set_option('display.max_rows' , None)
pd.set_option('display.max_columns' , None)
print(checking_scaled)
```
