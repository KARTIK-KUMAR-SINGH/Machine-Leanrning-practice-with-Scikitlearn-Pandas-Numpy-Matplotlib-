```python
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("Data.csv")
print (df.head())
```

```python
x = df.iloc[ : , 0 : 1]
y = df['Purchased']
```

```python
print(df['Country'].unique())
print(df['Purchased'].unique())
```

```python
x_train , x_test , y_train , y_test = train_test_split(x , y)
print(x_train)
```

```python
model1 = OrdinalEncoder(categories=[['France','Spain','Germany']])
model1.fit(x_train)
x_train = model1.transform(x_train)
x_test = model1.transform(x_test)
print(x_train)
```

```python
model2 = LabelEncoder()
model2.fit(y_train)
y_train = model2.transform(y_train)
y_test = model2.transform(y_test)
print(y_train)
```
