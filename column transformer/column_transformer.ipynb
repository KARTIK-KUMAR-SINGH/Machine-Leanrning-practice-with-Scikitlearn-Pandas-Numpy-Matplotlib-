```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("covid_toy.csv")
pd.set_option('display.max_rows' , None)
df.head()
df.tail()
```

```python
df.isnull().sum()
```

```python
x = df.drop(columns = ['has_covid'])
y = df['has_covid']
```

```python
print(x)
```

```python
print(y)
```

```python
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size = 0.2)
```

NORMAL METHOD

Applysing Simple Imputer

```python
si = SimpleImputer()
x_train_fever = si.fit_transform(x_train[['fever']])
x_test_fever = si.transform(x_test[['fever']])
print(x_train_fever.shape)
print(x_test_fever.shape)
```

Applying Ordinal Encoding

```python
x_train
```

```python
oe = OrdinalEncoder(categories = [['Mild' , 'Strong']])
x_train_cough = oe.fit_transform(x_train[['cough']])
x_test_cough = oe.transform(x_test[['cough']])

print(x_train_cough.shape)
print(x_test_cough.shape)
print(x_train_cough)
```

One Hot Encoding

```python
print(x_train)
```

```python
oh = OneHotEncoder(drop = 'first' , sparse_output = False)
x_train_gender_city = oh.fit_transform(x_train[['gender' , 'city']])
x_test_gender_city = oh.transform(x_test[['gender' , 'city']])
print(x_train.shape)
print(x_train_gender_city.shape)
print(x_test_gender_city.shape)
```

```python
print(x_train_gender_city)
```

Extracting Age colloumn

```python
x_train_age = x_train[['age']].values
x_test_age = x_test[['age']].values
# x_train_age = x_train.drop(columns=['gender','fever','cough','city']).values
print(x_train_age)
```

```python
x_train_transformed = np.concatenate((x_train_age , x_train_fever , x_train_cough , x_train_gender_city) , axis = 1)
x_test_transformed = np.concatenate((x_test_age , x_test_fever , x_test_cough , x_test_gender_city) , axis = 1)
```

```python
print(x_train_transformed)
```

```python
print(x_train_transformed.shape)
```

```python
print(x_test_transformed)
```

Using Column Transfomer

```python
transformer = ColumnTransformer(transformers =[
    ('tnf1' , SimpleImputer(),['fever']),
    ('tnf2' , OrdinalEncoder(categories = [['Mild' , 'Strong']]) , ['cough']),
    ('tnf3' , OneHotEncoder(drop = 'first' , sparse_output = False) , ['gender' , 'city'])
],remainder = 'passthrough')
```

```python
x_train_transformed_column_transformer = transformer.fit_transform(x_train)
x_test_transformed_column_transformer = transformer.transform(x_test)
print(x_train_transformed_column_transformer.shape)
print(x_test_transformed_column_transformer.shape)
```
